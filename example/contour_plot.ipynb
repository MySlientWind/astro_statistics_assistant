```python
import numpy as np
import matplotlib.pyplot as plt
from asa.dataset import Dataset
```

```python
x = np.random.normal(size=1000)
y = x**2 + np.random.normal(size=1000)
z = np.log(np.abs(x + y)) + np.random.normal(size=1000)
```

```python
data = np.array([x, y, z, x > 0, np.abs(y), y**2]).T
```

```python
ranges = {'x': [-3, 3]}
```

```python
dataset = Dataset(data, ['x', 'y', 'z', 'xg0', 'absy', 'y2'], ['x_label', 'y_label', 'z_label', 'xg0_label', 'absy_label', 'y2_label'], ranges=ranges)
```

```python
dataset.contour(['x', 'y'], 'z', subplots_kwargs={'figshape': (1, 2)})
```

```python
dataset.contour(['x', 'y'],
                'z',
                subplots_kwargs={'figshape': (1, 2)},
                levels=7) 
```

```python
dataset.contour(['x', 'y'],
                'z',
                subplots_kwargs={'figshape': (1, 2)},
                levels=[0.2, 0.5, 0.9])
```

```python
dataset.contour(['x', 'y'], 'z', subplots_kwargs={'figshape': (1, 2)}, range=None)
```
